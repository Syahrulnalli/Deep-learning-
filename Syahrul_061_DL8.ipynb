# Project 8: Gradient Descent Implementation

**Objective:** Implement a simple Gradient Descent algorithm to understand how model parameters are updated and how prediction error decreases during training.

## Task 1: Initialize Model Parameters

A simple linear dataset is used. The model is:

**y = wx + b**

where `w` is the weight and `b` is the bias.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simple dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

print("Input values:", X)
print("Actual values:", y)
```

```python
# Initialize parameters
w = 0.0
b = 0.0
learning_rate = 0.01
iterations = 1000

loss_history = []

print("Initial Weight:", w)
print("Initial Bias:", b)
print("Learning Rate:", learning_rate)
print("Iterations:", iterations)
```

## Task 2: Implement Gradient Descent

Gradient Descent repeatedly calculates the prediction error, finds the gradients, and updates the weight and bias to reduce the loss.

```python
n = len(X)

for i in range(iterations):
    y_pred = w * X + b

    # Mean Squared Error loss
    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    # Calculate gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Final Weight:", w)
print("Final Bias:", b)
print("Final Loss:", loss_history[-1])
```

```python
print("Initial Loss:", loss_history[0])
print("Final Loss:", loss_history[-1])
print("Loss Reduction:", loss_history[0] - loss_history[-1])
```

## Task 3: Display and Analyze Loss Reduction

```python
plt.figure(figsize=(8, 5))
plt.plot(loss_history)
plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Loss Reduction During Gradient Descent")
plt.grid(True)
plt.show()
```

### Prediction Graph

The trained model is compared with the original data.

```python
y_pred = w * X + b

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, label="Predicted Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent Prediction")
plt.legend()
plt.grid(True)
plt.show()
```

## Final Result

The Gradient Descent algorithm was successfully implemented using Python. The model parameters were updated iteratively, and the loss decreased over multiple iterations. This demonstrates how Gradient Descent reduces prediction error during model training.

## Conclusion

Gradient Descent is an optimization algorithm used to reduce the loss of a model. In this experiment, the weight and bias were updated step by step. As the number of iterations increased, the loss became smaller and the predictions became closer to the actual values.
