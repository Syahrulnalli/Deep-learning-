# Project 5: Activation Functions Using Python

## Task 1: Implement Activation Functions

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)
```

## Task 2: Generate Outputs

```python
x = np.array([-3, -2, -1, 0, 1, 2, 3])

print("Input:", x)
print("Sigmoid:", sigmoid(x))
print("ReLU:", relu(x))
print("Tanh:", tanh(x))
```

## Task 3: Visualize and Compare

```python
x = np.linspace(-5, 5, 100)

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, tanh(x), label="Tanh")
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Activation Functions")
plt.legend()
plt.grid(True)
plt.show()
```

## Result
Sigmoid gives values between 0 and 1. ReLU gives 0 for negative values and keeps positive values unchanged. Tanh gives values between -1 and 1.
